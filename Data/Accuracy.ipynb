```python
import pandas as pd
from sklearn.metrics import accuracy_score
```

```python
rows = [10, 50, 100, 150, 200, 250]
types = ["SAS", "MAS"]
opening = "Output with"
folders = ["Gemini 2.5 Flash", "GPT-4.1 2025-04-14"]
```

```python
for type in types:
    for r in rows:
        file_true = f"./Test/res_{r}_with_typos.csv"
        for folder in folders:
            file_output = f"./{folder}/NUS-SRI-2025-TEST - Sheet1 {r} {type}.csv"

            df_true = pd.read_csv(file_true, dtype=str).fillna("")
            df_output = pd.read_csv(file_output, dtype=str).fillna("")

            df_merged = pd.merge(
                df_true, df_output,
                on="NISN Siswa", how="outer",
                suffixes=("_true", "_output")
            )

            y_true, y_output = [], []

            for col in df_true.columns:
                if col == "NISN Siswa":
                    continue

                y_true.extend(df_merged[f"{col}_true"].astype(str).tolist())
                y_output.extend(
                    df_merged[f"{col}_output"].astype(str).tolist())

            print(
                f"Accuracy for {folder}, {r} rows, and {type} type: {accuracy_score(y_true, y_output)}")
```
